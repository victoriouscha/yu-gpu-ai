```python
# ==========================================
# 03. Gemma GPU LLM Practice
# Step 1. 실행 환경 확인
# ==========================================

import torch

print("PyTorch version :", torch.__version__)
print("CUDA available  :", torch.cuda.is_available())

if torch.cuda.is_available():
    print("GPU             :", torch.cuda.get_device_name(0))
else:
    print("GPU             : Not available (CPU mode)")
```

```python
# ==========================================
# Step 5-1. Gemma 모델 저장 경로 설정
# ==========================================

import os

cache_dir = "/root/hf-cache"
os.makedirs(cache_dir, exist_ok=True)

print("Cache directory :", cache_dir)
```

```python
# ==========================================
# Step 5-2. Gemma 3 4B BF16 모델 로드
# ==========================================

import torch
from transformers import AutoTokenizer, AutoModelForCausalLM

model_id = "google/gemma-3-4b-it"

tokenizer = AutoTokenizer.from_pretrained(
    model_id,
    cache_dir=cache_dir
)

torch.cuda.empty_cache()
torch.cuda.reset_peak_memory_stats()

model = AutoModelForCausalLM.from_pretrained(
    model_id,
    dtype=torch.bfloat16,
    device_map="auto",
    cache_dir=cache_dir
)

print("\n===== Gemma 모델 로드 완료 =====")
print("Model :", model_id)
print("GPU :", torch.cuda.get_device_name(0))
print(
    "GPU Memory Allocated :",
    round(torch.cuda.memory_allocated() / 1024**3, 2),
    "GB"
)
print(
    "GPU Memory Peak :",
    round(torch.cuda.max_memory_allocated() / 1024**3, 2),
    "GB"
)
```

```python
# ==========================================
# Step 6. Gemma 3 4B 추론 성능 측정
# ==========================================

import time
import torch

prompt = "생성형 AI와 기존 인공지능의 차이를 정확히 세 문장으로 설명해줘."

messages = [
    {"role": "user", "content": [{"type": "text", "text": prompt}]}
]

inputs = tokenizer.apply_chat_template(
    messages,
    add_generation_prompt=True,
    tokenize=True,
    return_dict=True,
    return_tensors="pt"
).to(model.device)

# Warm-up
with torch.inference_mode():
    _ = model.generate(
        **inputs,
        max_new_tokens=10,
        do_sample=False
    )

torch.cuda.synchronize()
torch.cuda.reset_peak_memory_stats()

start = time.perf_counter()

with torch.inference_mode():
    outputs = model.generate(
        **inputs,
        max_new_tokens=90,
        do_sample=False
    )

torch.cuda.synchronize()
elapsed = time.perf_counter() - start

input_tokens = inputs["input_ids"].shape[1]
generated_tokens = outputs.shape[1] - input_tokens
tokens_per_sec = generated_tokens / elapsed

response = tokenizer.decode(
    outputs[0][input_tokens:],
    skip_special_tokens=True
)

print("===== Gemma 3 4B 추론 결과 =====")
print(response)

print("\n===== 성능 측정 =====")
print("GPU :", torch.cuda.get_device_name(0))
print("Precision : BF16")
print("Generated tokens :", generated_tokens)
print("Generation time :", round(elapsed, 3), "sec")
print("Speed :", round(tokens_per_sec, 2), "tokens/sec")
print(
    "GPU Memory Allocated :",
    round(torch.cuda.memory_allocated() / 1024**3, 2),
    "GB"
)
print(
    "GPU Memory Peak :",
    round(torch.cuda.max_memory_allocated() / 1024**3, 2),
    "GB"
)
```
