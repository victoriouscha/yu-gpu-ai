```python
import requests

BASE_URL = "http://127.0.0.1:8000"

question = "파이썬의 리스트와 튜플의 차이를 쉽게 설명해줘."

response = requests.post(
    BASE_URL + "/agent",
    json={"query": question},
    timeout=240
)

data = response.json()

print("🤖 AI 답변:", data.get("final_answer"))
print("🔧 사용한 Tool:", data.get("tool_name"))
print("✅ 실행 상태:", data.get("status"))
```
