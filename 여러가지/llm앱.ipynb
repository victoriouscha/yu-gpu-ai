```python
import sys

sys.path.insert(0, "/workspace/05_agent_web_mvp")

from app.llm import generate_llm

job = input("🎯 희망 취업 직무를 입력하세요: ")

question = f"{job} 직무에 지원한 대학생에게 면접 질문 3개를 만들어줘. 질문만 간단히 제시해줘."

result = generate_llm(
    [{"role": "user", "content": question}],
    max_new_tokens=200
)

print("\n🎤 AI 모의면접 질문")
print(result)
```

```python

# 앞 단계에서 불러온 generate_llm() 함수를 그대로 사용합니다.

print("🎤 AI 면접 답변 첨삭 프로그램")

# ① 면접 질문 입력
interview_question = input("면접 질문을 입력하세요: ")

# ② 학생 답변 입력
student_answer = input("나의 답변을 입력하세요: ")

# ③ AI에게 첨삭 요청
prompt = f"""
당신은 {job} 직무의 전문 면접관입니다.

[면접 질문]
{interview_question}

[학생 답변]
{student_answer}

다음 세 가지 항목으로 한국어 첨삭을 해주세요.

1. ✅ 잘한 점: 한 문장
2. 💡 개선할 점: 한 문장
3. ✨ 개선된 답변 예시: 3문장 이내

학생이 실제로 경험하지 않은 경력이나 성과를
임의로 만들어 넣지 마세요.
"""

# ④ Qwen3-8B 실행
if interview_question.strip() and student_answer.strip():
    feedback = generate_llm(
        [{"role": "user", "content": prompt}],
        max_new_tokens=400
    )

    # ⑤ 첨삭 결과 출력
    print("\n" + "=" * 45)
    print("🤖 AI 면접관의 첨삭 결과")
    print("=" * 45)
    print(feedback)
else:
    print("⚠️ 면접 질문과 답변을 모두 입력하세요.")
```
